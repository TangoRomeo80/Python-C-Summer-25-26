```python
# importing necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/refs/heads/master/titanic.csv"
df_raw = pd.read_csv(url)
```

```python
df = df_raw.copy()
```

```python
df.describe(include='all') # Provides summary statistics for all columns, including categorical columns
```

```python
# Plotting the distribution of the 'fare' column
plt.figure(figsize=(10, 6))
plt.hist(df['fare'], bins=30, edgecolor='black')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.title('Distribution of Fare')
plt.show()
```

```python
print('Mean fare:', df['fare'].mean())
print('Median fare:', df['fare'].median())
```

```python
# Categorical variables description
#print(df['embark_town'].value_counts()) # Counts the occurrences of each unique value in the 'pclass' column
print(df['embark_town'].value_counts(normalize=True)) # Counts the occurrences of each unique value in the 'pclass' column and normalizes it to show proportions
# print(df['embark_town'].value_counts(dropna=False)) # Counts the occurrences of each unique value in the 'pclass' column, including NaN values
```

```python
# print(df.isna()) # Shows if a value is missing or not
print(df.isna().sum()) # Shows the total number of missing values in each column
```

```python
# Visualize missing values using a bar chart
missing_values = df.isna().sum()

plt.figure(figsize=(10, 6))
missing_values.plot(kind='bar', color='skyblue')
plt.title('Missing Values in Each Column')
plt.xlabel('Columns')
plt.ylabel('Number of Missing Values')
plt.xticks(rotation=45)
plt.show()
```

```python
print(df.groupby(['pclass'])['age'].median()) # Group the DataFrame by 'pclass' and calculate the median age for each class
```
